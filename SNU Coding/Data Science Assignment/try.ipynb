```python
import random
from PIL import Image, ImageDraw

def generate_fractal_image(width, height, num_iterations):
    # 1. Initialize a coordinate (x0, y0) and put a dot at (x0, y0).
    x, y = 500, 500

    # Create a new blank black image
    img = Image.new('RGB', (width, height), 'black')
    draw = ImageDraw.Draw(img)

    # Put the initial dot
    draw.point((int(x), int(y)), 'white')

    # 5. Repeat the process a large number of times and generate the image.
    for _ in range(num_iterations):
        # 2. Generate a random no between 1 and 4.
        random_no = random.randint(1, 4)

        if random_no == 1:
            x_new = 0.8 * x + 0.1 * width
            y_new = 0.8 * y + 0.04 * height
        elif random_no == 2:
            x_new = 0.5 * x + 0.25 * width
            y_new = 0.5 * y + 0.4 * height
        elif random_no == 3:
            x_new = 0.355 * (x - y) + 0.266 * width
            y_new = 0.355 * (x + y) + 0.098 * height
        elif random_no == 4:
            x_new = 0.355 * (x + y) + 0.378 * width
            y_new = 0.355 * (y - x) + 0.434 * height
        
        # Update the coordinates for the next iteration
        x, y = x_new, y_new

        # 4. Put a dot at new coordinate (x, y).
        if 0 <= x < width and 0 <= y < height:
            draw.point((int(x), int(y)), 'white')

    # Save the generated image
    img.save('generated_image.png')
    print("Image saved as generated_image.png")

# Set image parameters
image_width = 1000
image_height = 1000
iterations = 100000  # A large number of iterations to get a good result

# Run the function to generate the image
generate_fractal_image(image_width, image_height, iterations)
```
